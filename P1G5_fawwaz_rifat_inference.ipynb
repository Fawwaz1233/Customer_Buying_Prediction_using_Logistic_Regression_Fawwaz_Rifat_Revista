```python
#import library

import pickle
import json
import pandas as pd
import numpy as np
```

```python
#Load model

with open('list_cat_cols.txt', 'r') as file_1:
  list_cat_col = json.load(file_1)

with open('list_num_cols.txt', 'r') as file_2:
  list_num_col = json.load(file_2)

with open('encoder.pkl', 'rb') as file_3:
  model_encoder = pickle.load(file_3)

with open('scaler.pkl', 'rb') as file_4:
  model_scaler = pickle.load(file_4)

with open('model_lin_reg.pkl', 'rb') as file_5:
  model_log_reg = pickle.load(file_5)
```

```python
data_inf = {
    'buying' : 0,
    'time_on_site' : 25,
    'pageviews' : 1.0,
    'latest_ecommerce_progress' : '0',
    'bounces' : '0',
    'medium' : 'referral',
    'channelGrouping' : 'Social',
    'deviceCategory' : 'mobile'
}

data_inf = pd.DataFrame([data_inf])
data_inf
```

```python
#Split data numerik dan kategorik

data_inf_num = data_inf[list_num_col]
data_inf_cat = data_inf[list_cat_col]
data_inf_num
```

```python
# Feature scaling and encoding

data_inf_num_scaled = model_scaler.fit_transform(data_inf_num)
data_inf_cat_encoded = model_encoder.fit_transform(data_inf_cat)
data_inf_final = np.concatenate([data_inf_num_scaled, data_inf_cat_encoded], axis = 1)
```

```python
# Prediksi menggunakan nilai Regresi Logistik
y_pred_inf = model_log_reg.predict(data_inf_final)
y_pred_inf
```
